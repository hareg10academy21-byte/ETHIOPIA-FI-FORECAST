```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
access_data={

    "Year":[

        2011,
        2014,
        2017,
        2021,
        2024

    ],

    "Ownership":[

        14,
        22,
        35,
        46,
        49

    ]

}
```

```python
access=pd.DataFrame(access_data)

access
```

```python
X=access[["Year"]]

y=access["Ownership"]
```

```python
model=LinearRegression()

model.fit(X,y)
```

```python
future_years=pd.DataFrame(

        {

            "Year":[

                2025,
                2026,
                2027

            ]

        }

)
```

```python
forecast=model.predict(

        future_years

)
```

```python
forecast
```

```python
future_years["Forecast"]=forecast

future_years
```

```python
plt.figure(figsize=(8,5))

plt.plot(

    access["Year"],

    access["Ownership"],

    marker="o",

    label="Historical"

)


plt.plot(

    future_years["Year"],

    future_years["Forecast"],

    marker="o",

    label="Forecast"

)


plt.title(

"Account Ownership Forecast (2025-2027)"

)

plt.xlabel(

"Year"

)

plt.ylabel(

"Ownership (%)"

)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
future_years["Lower"]=(

future_years["Forecast"]-3

)

future_years["Upper"]=(

future_years["Forecast"]+3

)
```

```python
future_years
```

```python
plt.figure(figsize=(8,5))


plt.plot(

    future_years["Year"],

    future_years["Forecast"],

    marker="o"

)


plt.fill_between(

    future_years["Year"],

    future_years["Lower"],

    future_years["Upper"],

    alpha=0.3

)


plt.title(

"Forecast with Confidence Interval"

)

plt.xlabel(

"Year"

)

plt.ylabel(

"Ownership (%)"

)

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
future_years["Optimistic"]=(
future_years["Forecast"]+5
)


future_years["Base"]=(
future_years["Forecast"]
)


future_years["Pessimistic"]=(
future_years["Forecast"]-5
)
```

```python
future_years
```

```python
plt.figure(figsize=(10,6))


plt.plot(

future_years["Year"],

future_years["Optimistic"],

marker="o",

label="Optimistic"

)



plt.plot(

future_years["Year"],

future_years["Base"],

marker="o",

label="Base"

)



plt.plot(

future_years["Year"],

future_years["Pessimistic"],

marker="o",

label="Pessimistic"

)



plt.title(

"Scenario Forecasts (2025-2027)"

)


plt.xlabel(

"Year"

)


plt.ylabel(

"Account Ownership (%)"

)


plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()
```
