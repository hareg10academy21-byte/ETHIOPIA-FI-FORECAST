```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


# Historical Digital Payment Adoption

usage = pd.DataFrame({

"Year":[2011,2014,2017,2021,2024],

"Usage":[5,8,15,28,35]

})


X = usage[["Year"]]

y = usage["Usage"]


model = LinearRegression()

model.fit(X,y)



future = pd.DataFrame({

"Year":[2025,2026,2027]

})


forecast = model.predict(future)


forecast
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


usage_data = pd.DataFrame({

    "year":[2014,2017,2021,2024],

    "usage":[
        5,
        10,
        20,
        35
    ]

})


usage_data
```

```python
X = usage_data[["year"]]

y = usage_data["usage"]


usage_model = LinearRegression()


usage_model.fit(
    X,
    y
)
```

```python
future_usage = pd.DataFrame({

    "year":[2025,2026,2027]

})


future_usage
```

```python
usage_prediction = usage_model.predict(
    future_usage
)


usage_prediction
```

```python
usage_forecast_table = pd.DataFrame({

    "Year":[2025,2026,2027],

    "Digital Payment Usage Forecast (%)":
    usage_prediction.round(2)

})


usage_forecast_table
```

```python
plt.figure(figsize=(10,6))


plt.plot(
usage_data["year"],
usage_data["usage"],
marker="o",
label="Historical Usage"
)


plt.plot(
future_usage["year"],
usage_prediction,
marker="o",
label="Forecast Usage"
)


plt.title(
"Ethiopia Digital Payment Usage Forecast 2025-2027"
)


plt.xlabel(
"Year"
)


plt.ylabel(
"Digital Payment Usage (%)"
)


plt.legend()


plt.grid()


plt.show()
```

```python
plt.figure(figsize=(10,6))


plt.plot(
usage["Year"],
usage["Usage"],
marker="o",
label="Historical"
)


plt.plot(
future["Year"],
forecast,
marker="o",
label="Forecast"
)


plt.title(
"Digital Payment Adoption Forecast Ethiopia 2025-2027"
)


plt.xlabel("Year")


plt.ylabel(
"Adults Using Digital Payments (%)"
)


plt.legend()


plt.grid(True)


plt.show()
```

```python
future_years["Lower"]=(

future_years["Forecast"]-3

)

future_years["Upper"]=(

future_years["Forecast"]+3

)
```

```python
future_years
```

```python
plt.figure(figsize=(8,5))


plt.plot(

    future_years["Year"],

    future_years["Forecast"],

    marker="o"

)


plt.fill_between(

    future_years["Year"],

    future_years["Lower"],

    future_years["Upper"],

    alpha=0.3

)


plt.title(

"Forecast with Confidence Interval"

)

plt.xlabel(

"Year"

)

plt.ylabel(

"Ownership (%)"

)

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
future_years["Optimistic"]=(
future_years["Forecast"]+5
)


future_years["Base"]=(
future_years["Forecast"]
)


future_years["Pessimistic"]=(
future_years["Forecast"]-5
)
```
